```python
import pandas as pd

# Read the file (tab-separated)
url ="https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"

df = pd.read_csv(url, sep ='\t')

```

```python
df.head(10)
```

```python
df.shape[0]
```

```python
df.info()
```

```python
df.shape[1]
```

```python
df.columns
```

```python
# which was the most ordered item
i = df.groupby('item_name')
i = i.sum()
i = i.sort_values(['quantity'], ascending=False)
i.head(1)
```

```python
# what was the most ordered item in the choice description
i = df.groupby('choice_description').sum()
i = i.sort_values(['quantity'], ascending=False)
i.head(1)
```
