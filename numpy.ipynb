### Check Number of Dimensions in an NumPy Array?

```python
# check Number of dimensions in an NumPy Array?
import numpy as np

a = np.array(42)
b = np.array([1,2,3,4,5])
c = np.array([[1,2,3],[4,5,6]])
d = np.array([[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

Filtering of an array

- Getting out some elements from existing array and create new array out of them is called filtering.
- In NumPy, you filter any array ising a boolean index list.
- if value at an index is True that element is contained in the filtered array,
- if the value at that index is False, then the element is excluded from the filtered array.

```python
arr = np.array([41, 42, 43, 44])
x = [False, True, True, False]

newarr = arr[x]
print(newarr)
```

```python
# Create a filter array that will return only values higher than 42
arr = np.array([41, 42, 43, 44])
x = [False, False, True, True]

newarr = arr[x]
print(newarr)
```

```python
import numpy as np

arr = np.array([41, 42, 43, 44])

# Create a filter using a condition
filter = arr > 42

newarr = arr[filter]
print(newarr)
```
