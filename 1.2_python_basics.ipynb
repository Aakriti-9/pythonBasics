## Conditional statement in python ( if, elif, else)

```python

## if <condition>: statements
age = 21
if age>=18:
  print("you are adult")
```

```python
age = 21
if age>=20:
  print("you are adult")
else:
  print("you are teen")
```

```python
num = int (input("Enter a number"))
if num > 0:
  print("positive number")
  if num % 2 == 0 :
    print("Even number")
  else:
    print("Odd Number")
elif num == 0:
  print("Zero")
else:
  print("Negative number")


```

```python
year = int(input("Enter a year"))
if year % 4 == 0:
  print("Leap Year")
  if year % 100 == 0:
    print("Not a leap Year.")
    if year % 400 == 0:
      print("Leap Year")
else:
  print("Not a leap Year")
```

```python
age = int(input("Enter your age"))
is_student = input("Are you a student (yes/no)")
if age < 5:
  print("The ticket price is free")
elif age>=5 and  age<=12:
  print("The ticket price is $10")
elif age>=13 and age<=17:
  if is_student == "no":
    print("The ticket price is $15")
  elif is_student == "yes":
      print("The ticket price is $12")
elif age>=18 and age<64:
  if is_student == "no":
    print("The ticket price is $25")
  elif is_student == "yes":
      print("The ticket price is $18")
else:
  print("The ticket price is $20")
```

```python
salary = int(input("Enter your salary"))
pr = float(input("Enter your performance rating(1.0 to 5.0)"))
year = int(input("Enter your year of service"))
if pr >= 4.5 and pr <= 5.0:
  if year > 10 :
    bonus_percent = 20
  elif year > 5 :
     bonus_percent = 15
  elif year <= 5 :
    bonus_percent = 10
  else:
   bonus_percent = 0

if pr >= 3.5 and pr <= 4.49:
  if year > 10 :
    bonus_percent = 15
  elif year > 5 :
     bonus_percent = 10
  elif year <= 5 :
    bonus_percent = 5
  else:
   bonus_percent = 0

Bonus_amount = salary * (bonus_percent/100)
print("Bonus Amount:", Bonus_amount)
```
