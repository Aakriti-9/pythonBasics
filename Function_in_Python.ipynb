```python
## 2 Password strength checker
# condition for strong password
 # length should be 8 or >8
 # pw should contain at least one digit
 # atleast one letter should be uppercase and one should be lowercase
 # should use special character like  !@$%^

 # strong else week password.
```

```python
def pass_check(password):
    if len(password) >= 8:
        # checks the digit
        for i in range(len(password)):
            if password[i].isdigit():

                # checks the uppercase
                for j in range(len(password)):
                    if password[j].isupper():
                        # check special character
                                special = "!@#$%^&*()-_=+[]{};:'\",.<>?/\\|`~"
                                for m in range(len(password)):
                                    if password[m] in special:
                                        return ("Strong Password")
    return ("Weak Password")






```

```python
pass_check("Pransocool@99")
```

```python
pass_check("Pransocool")
```

## Homework

```python
def fac(num):
  factorial = 1
  for i in range(1, num+1):
    factorial = factorial * i
  return factorial

fac(5)
```

```python
fac(6)
```

```python
def palindrome(word):
  if word == word[::-1]:
    return True
  else:
    return False

palindrome("madam")
```

```python
def validate_email(email):
    if "@" in email and "." in email:
        return True
    else:
        return False

print(validate_email("test@example.com"))
print(validate_email("hello.com"))
print(validate_email("noatsymbol"))
```

## Lambda Function

- lambda function are small anonymus function defined using the keyword
'''lamda''' .
- It can have any number of arguments but only one expression.
- They are commonly used for short operations or arguments to higher order function


```python
## syntax
lambda arguments : Expression
```

```python
def add (a,b):
    return a+b
```

```python
add(2, 3)
```

```python
add = lambda a, b : a+b
print(add(2,3))
```

```python
## check whether the number is even or odd, using lambda function

num = lambda a : a % 2 == 0
if num(2):
  print("even")
else:
  print("odd")

```

```python
# even number

def even(num):
    if num % 2 == 0:
       return True
```
