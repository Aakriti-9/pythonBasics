 ## Nested Tuple with Nested list

```python
nested_tuple = ((1, 2, 3), ('a', 'b', 'c'), (True, False))
nested_tuple
```

```python
nested_tuple[1][2]
```

```python
nested_tuple[2][1]
```

```python
## iterating the elwments in a tuple
nested_tuple
```

```python
for sub_tuple in nested_tuple:
  print(sub_tuple)
```

```python
for sub_tuple in nested_tuple:
  for element in sub_tuple:
      print(element, end=" ")
  print()
```
