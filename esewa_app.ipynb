```python
import random
wallet_registry = {} # dictionary to store all the wallets
```

```python
class DigitalWallet:

    def __init__(self, user_name, password, initial_balance=0):
        self.owner = user_name
        self.wallet_id = f"NP-{random.randint(10000, 99999)}"
        self._balance = initial_balance #protected attribute
        self.history = [] # list to store all the transactions
        self.max_limit = 50000 # Normal wallet limit: Rs 50000
        self._password = password
        self._authenticated = False

     # Register wallet automatically
        wallet_registry[self.owner] = self

    def login(self, user_name, password):
        if self.owner == user_name and self._password == password:
            self._authenticated = True
            print(f"Login successful for {self.owner}")
        else:
            print("Login failed! Wrong username or password.")

    def LoadMoney(self, amount):
        if not self._authenticated:
            print("Please login first!")
            return

        if amount > self.max_limit:
            print(f"You cannot load more than Rs. {self.max_limit}")
        elif amount <= 0:
            print("Invalid amount")
        else:
            self._balance += amount
            self.history.append(f"Loaded Rs.{amount}")
            print(f"Rs.{amount} loaded successfully.")

    def Payment(self, amount, description):
        if not self._authenticated:
            print("Please login first!")
            return

        if amount > self.max_limit:
            print(f"You cannot pay more than Rs. {self.max_limit}")
        elif amount <= 0:
            print("Invalid amount")
        elif amount > self._balance:
            print("Insufficient Balance")
        else:
            self._balance -= amount
            self.history.append(
                f"Paid Rs.{amount} for {description}"
            )
            print("Payment successful.")

    def TransferMoney(self, amount, receiver_name):
        if not self._authenticated:
            print("Please login first!")
            return

        if amount > self.max_limit:
            print(f"You cannot transfer more than Rs. {self.max_limit}")
            return

        if amount <= 0:
            print("Invalid amount")
            return

        if amount > self._balance:
            print("Insufficient Balance")
            return

        receiver = wallet_registry.get(receiver_name)

        if receiver:
            self._balance -= amount
            receiver._balance += amount

            self.history.append(
                f"Transferred Rs.{amount} to {receiver.owner}"
            )

            receiver.history.append(
                f"Received Rs.{amount} from {self.owner}"
            )

            print("Transfer successful.")
        else:
            print("Receiver not found")


    def check_status(self):
        print("\n----- Wallet Status -----")
        print(f"Owner: {self.owner}")
        print(f"Wallet ID: {self.wallet_id}")
        print(f"Balance: Rs.{self._balance}")

        print("Transaction History:")

        if len(self.history) == 0:
            print("No transactions yet.")
        else:
            for transaction in self.history:
                print(transaction)



```

```python
# need to inherit from Parent class
class PremiumWallet(DigitalWallet):

    def __init__(self, user_name, password, initial_balance=0):
      super().__init__(user_name, password, initial_balance)
      self.max_limit = 500000 # Premium wallet limit: Rs 500000
      self.cashback_rate = 0.02 # 2% cash back on every bill payment

     # Method Overriding
    def Payment(self, amount, description):

        if not self._authenticated:
            print("Please login first!")
            return

        if amount > self.max_limit:
            print(f"You cannot pay more than Rs. {self.max_limit}")

        elif amount <= 0:
            print("Invalid amount")

        elif amount > self._balance:
            print("Insufficient Balance")

        else:
            self._balance -= amount

            cashback = amount * self.cashback_rate

            self._balance += cashback

            self.history.append(
                f"Paid Rs.{amount} for {description}"
            )

            self.history.append(
                f"Cashback Rs.{cashback}"
            )

            print("Payment successful (Premium User).")
            print(f"Cashback received: Rs.{cashback}")

    # Method Overriding
    def TransferMoney(self, amount, receiver_name):

        if not self._authenticated:
            print("Please login first!")
            return

        if amount > self.max_limit:
            print(f"You cannot transfer more than Rs. {self.max_limit}")
            return

        if amount <= 0:
            print("Invalid amount")
            return

        if amount > self._balance:
            print("Insufficient Balance")
            return

        receiver = wallet_registry.get(receiver_name)

        if receiver:

            self._balance -= amount
            receiver._balance += amount

            cashback = amount * self.cashback_rate
            self._balance += cashback

            self.history.append(
                f"Transferred Rs.{amount} to {receiver.owner}"
            )

            self.history.append(
                f"Cashback Rs.{cashback}"
            )

            receiver.history.append(
                f"Received Rs.{amount} from {self.owner}"
            )

            print("Transfer successful (Premium User).")
            print(f"Cashback received: Rs.{cashback}")

        else:
            print("Receiver not found")

    def check_status(self):
        super().check_status()
        print(f"Premium, Cashback Rate: {self.cashback_rate * 100}%")
```

```python
# Create wallets
wallet1 = DigitalWallet("Aakriti Kandel", "Aakriti333", 10000)
wallet2 = PremiumWallet("Manisha Kandel", "Manisha899", 50000)

# Login first (mandatory)
wallet1.login("Aakriti Kandel", "Aakriti333")
wallet2.login("Manisha Kandel", "Manisha899")

# Perform operations
wallet1.LoadMoney(5000)
wallet1.Payment(2000, "Groceries")

wallet2.Payment(10000, "Shopping")
wallet2.TransferMoney(5000, "Aakriti Kandel")

# Check status
wallet1.check_status()
wallet2.check_status()
```
