# Exception Handeling

- understanding Exception
   - run time error handelling.
   - if run time, logocal error aayo vane pani, it makes sure that, the correction actions will be taken without stopping execution of the program.

Exception
- define an exception as an event, that disrupt the normal flow of a program.
- It occurs when an error is encountered during the program execution.

Common exception includes
  - ZeroDivisionError: Divide by Zero.
  - FileNotFoundError: file not found
  - ValueError: Invalid value
  - TypeError: Invalid type

To handle exception ⁉
- try
- except
- else
- finally

```python
# except error
try:
  #statement
  a = b
except:
  # statement
  print("the variable not defined")
```

```python
# try, eccept and else block

try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError as ex:
    print(ex)
    print("Dominator should always be greater than 0")
except NameError as ex:
    print(ex)
    print("<<<<<<< Main exception is caught here >>>>>>>")
else:
    print("No exception")
    print(result)
```
