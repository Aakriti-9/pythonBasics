### Multiple methods

- A class can have multiple methods that work together

````python
class Playlist:
  def __init__(self, name):
    self.name = name
    self.songs = []

  def add_song(self, song):
      self.songs.append(song)
      print(f"Added a song:{song}")

  def remove_song(self, song):
      if song in self.songs:
         self.songs.remove(song)
         print(f"Removed a song:{song}")

  def show_songs(self):
      print(f"Playlist Name: {self.name}")
      for song in self.songs:
        print(song)

my_playlist = Playlist("Favourites")
my_playlist.add_song("Song 1")
my_playlist.add_song("Song 2")
my_playlist.add_song("Song 3")
my_playlist.show_songs()
````

````python
my_playlist.remove_song("Song 3")
my_playlist.show_songs()
````

````python
class Mother: # parent class 1
  mothername = ""

  def mother(self):
    print(self.mothername)

class Father: # parent class 2
  fathername = ""

  def father(self):
    print(self.fathername)


class Son(Mother, Father): # child class
  def parents(self):
    print("Father :", self.fathername)
    print("Mother :", self.mothername)

s1 = Son()
s1.fathername = "Ram"
s1.mothername = "Sita"
s1.parents()
````

Multilevel Inheritance

image.png

````python
class Grandfather:
  def __init__(self, grandfathername):
    self.grandfathername = grandfathername

class Father(Grandfather):
    def __init__(self, fathername, grandfathername):
      self.fathername = fathername
      Grandfather.__init__(self, grandfathername)

class Son(Father):
    def __init__(self, sonname, fathername, grandfathername):
      self.sonname = sonname
      Father.__init__(self, fathername, grandfathername)

    def print_name(self):
      print('Grandfather name :', self.grandfathername)
      print("Father name :", self.fathername)
      print("Son name :", self.sonname)

s1 = Son('Prince', 'Rampal', 'Lal mani')
print(s1.grandfathername)
s1.print_name()
````

Important notes:

- Son inherits from father, and father inherits from Grandfather

- Each constructor passes values up the inheritance chain using explicit constructor calls

### super() function
- super() function that will make the child class inherit all the methods and properties from its parents class

- it os commonly used in the child class's ``init()`` method to intialize inherited attributes.

````python
#Animal

class Animal:
    def __init__(self, name):
        self.name = name

    def info(self):
        print("Animal Name:", self.name)

#child class

class Dog(Animal):
    def __init__(self, name, breed):
       # call contructor based upon python's method resolution order (MRO)
       super(). __init__(name)
       self.breed = breed

    def details(self):
        print(self.name, "is a", self.breed)

d = Dog("Khaire", "Bhusiya")
d.info()
d.details()
````

Takeaway:

- ```super()``` function is used inside ```__init__``` method of Dog to call the constructor of ```Animal``` and initialize inherited attribute (name)

- It ensures that parent class functionality is reused without needing to rewrite the code in the child class.

````python
# Overriding

class Animal:
    def sound(self):
        return "Some generic sound"

class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"

animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.sound())
````

 - Method Overriding

 Sound method behaves differently depending on whether the object is Dog or Cat or Animal, and this type of decision is made during the runtime. Thus, the dynamic nature makes Python particularly powerful for runtime Polymorphism,

Practice Question: Polymorphism with Functions and Methods

Create a base class Shape with a method area().

Then create two derived classes:

- Rectangle — takes width and height and calculates the area.
- Circle — takes radius and calculates the area.

Finally, create a function print_area(shape) that accepts any shape object and prints its area.

````python
class Shape:
    def area(self,):
        return "Area of different shapes"

class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

shapes = [Rectangle(5, 10), Circle(7), Shape()]
for shape in shapes:
    print(shape.area())
````
