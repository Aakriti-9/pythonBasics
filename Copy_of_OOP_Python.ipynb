Task:
- Create a class named Caculator and add following methods
- method add return a + b
- multiple return a * b
- divide return a / b
- subtraction return a - b

````python
class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def multiply(self):
      return self.num1 * self.num2

    def divide(self):
      return self.num1 / self.num2

    def subtract(self):
      return self.num2 - self.num1
````

````python
obj1 = Calculator(10,20)
print("Addition is ", obj1.add())
print("Multiplication is ",obj1.multiply())
print("Subtraction is ", obj1.subtract())
print("Division is ", obj1.divide())
````

### Multiple methods

- A class can have multiple methods that work together

````python
class Playlist:
  def __init__(self, name):
    self.name = name
    self.songs = []

  def add_song(self, song):
      self.songs.append(song)
      print(f"Added a song:{song}")

  def remove_song(self, song):
      if song in self.songs:
         self.songs.remove(song)
         print(f"Removed a song:{song}")

  def show_songs(self):
      print(f"Playlist Name: {self.name}")
      for song in self.songs:
        print(song)

my_playlist = Playlist("Favourites")
my_playlist.add_song("Song 1")
my_playlist.add_song("Song 2")
my_playlist.add_song("Song 3")
my_playlist.show_songs()
````

````python
my_playlist.remove_song("Song 3")
my_playlist.show_songs()
````

## Inheritance in python

Inheritance allows us to define a class that inherits all the methods and properties from another class

- parent class : class being inherited from, also call it as ***base class***

- child class: Class that inherits from another class, also call it as ***derived class***

````python
# any class can be a parent class, sp the syntax remains same  as above defination

class Person:
   def __init__(self, fname, lname):
      self.firstname = fname
      self.lastname = lname

   def printname(self):
      print(self.firstname, self.lastname)

# use the person class to create an object, and then execute the printname method

person1 = Person("Aakriti", "Kandel")
person1.printname()


````

````python
class Student(Person):
  pass
````

````python
person2 = Student("Mahesh", "Kandel")
person2.printname()
````
